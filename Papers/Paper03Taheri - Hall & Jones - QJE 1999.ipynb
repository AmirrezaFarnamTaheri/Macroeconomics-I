## Introduction

Hall and Jones suggest that, differences in capital accumulation, and output per worker can be explained by productivity, which itself is related to differences in social infrastructure.

They suggest that countries with better social infrastructures will have higher levels of output per worker because they support productive activities, encourage capital accumulation, and facilitate skill acquisition, invention, and technology transfer.

They focus on levels instead of rates of growth shich is important on some aspects such that it captures differences in long-run economic welfare more directly, as measured by the consumption of goods and services.

In other word, they are asking where is the steady-state of economy instead of when do we reach the steady state. This is important because we know that the convergance rates converge together across the countries.



## Methodology

They suggest that they might be reverse causality between social infrastracture, $s$, and human capital and output per worker, $y$.

To find the exogenous effect of $s$, they have analyzed the countries, which their infrastructure have been developed by european colonizer.

To identify these infrustructure they have used sevral instruments such as:

1. **Distance from the Equator:**

The countries closer to the equator have been less likely to been influenced by Western European colonial powers.

2. **Language Spoken:**

Another instrument discussed is the fraction of a country's population speaking one of the five primary Western European languages (English, French, German, Portuguese, and Spanish) as a mother tongue. This variable captures the cultural and linguistic legacy of Western European colonialism. The paragraph explains that countries where these languages are spoken as first languages are more likely to have been strongly influenced by Western European ideas, institutions, and infrastructure. 

3. **Frankel-Romer Predicted Trade Share:**

They have also used the Frankel-Romer predicted trade share variable is also used as an instrument. Their variable captures the economic integration of countries, which can indirectly influence social infrastructure. since economic integration is relevant with trade which can lead to the diffusion of ideas, technologies, and institutions.
