## $$\small\textbf{The Output, Employment, and Interest Rate Effects of Government Consumption}$$ $$\scriptsize\textbf{by S. Rao Aiyagari, Lawrence J. Christiano, and Martin Eichenbaum}$$

#### $\textbf{Introduction}$


This paper examines the effects of government consumption changes on employment, output, and interest rates within a modified neoclassical growth model framework.

#### $\textbf{Methodology}$


- The model is based on a competitive equilibrium allocation problem, where households maximize their utility from consumption and leisure, subject to budget and technological constraints.

- The authors use a variant of the neoclassical growth model, which includes elastic labor supply and government consumption, to analyze the effects of changes in government consumption on employment, output, and interest rates.

#### $\textbf{Results}$



- Both temporary and permanent increases in government consumption boost output and employment.

- The steady-state response of output to a permanent change in government consumption can exceed unity.

- Real interest rates rise in response to increased government consumption, regardless of its persistence.
