## $$\small\textbf{Can Technology Improvements Cause Productivity Slowdowns?}$$ $$\scriptsize\textbf{by Andreas Hornstein and Per Krusell - NBER 1996}$$

### $\textbf{Introduction and Main Question}$

The paper explores the counterintuitive idea that technological progress can lead to productivity slowdowns, challenging the conventional view that technological progress always leads to productivity improvements. The authors question whether increases in the rate of investment-specific technological change can cause decreases in measured productivity growth.



### $\textbf{Methodology and Identification}$


The authors use a two-sector vintage capital model to analyze the relationship between technological progress and productivity growth. They identify two channels through which technological progress can lead to productivity slowdowns: learning and unmeasured quality. The learning channel suggests that resources are devoted to new technologies where experience is low, leading to decreased productivity growth. The unmeasured quality channel implies that increases in capital input lead to declines in productivity growth if the unmeasured quality of final outputs depends significantly on capital input.


### $\textbf{Results}$


The results show that increases in the rate of investment-specific technological change can indeed lead to decreases in measured productivity growth through both channels. The authors document the recent productivity slowdown in the United States and elsewhere and provide evidence suggesting that an increase in the rate of investment-specific technological change may have occurred at about the same time as the slowdown began.


### $\textbf{Conclusion}$


The paper concludes that technological progress can, in fact, cause productivity slowdowns, challenging the conventional view. The findings have important implications for understanding the relationship between technological progress and productivity growth. The authors emphasize the importance of considering the learning curve associated with new technology adoption and the potential for quality mismeasurement when evaluating productivity data. The paper highlights the complex and sometimes counterintuitive relationship between technology improvements and economic performance.
