# Can Technology Improvements Cause Productivity Slowdowns?
**Authors:** Andreas Hornstein and Per Krusell (*NBER Macroeconomics Annual*, 1996)

---

## 1. Introduction & Research Problem
The paper examines the empirical paradox that major waves of technological progress can be accompanied by a protracted slowdown in measured productivity growth. Hornstein and Krusell address whether the acceleration in investment-specific technological change (embodied technical change) during the 1970s and 1980s directly accounts for the simultaneous productivity slump observed across industrialized economies.

## 2. Theoretical Framework: Two-Sector Vintage Capital Model
The authors construct a dynamic equilibrium model featuring:
- **Vintage Capital Structure:** New capital goods incorporate cutting-edge technical improvements, making older capital vintages economically obsolete.
- **The Learning-by-Doing Mechanism:** Operating new vintages requires acquiring specialized operational knowledge. During the transitional learning phase, resources are reallocated toward adaptation and restructuring, temporarily reducing current output per measured unit of input.
- **The Unmeasured Quality Mechanism:** Official statistical indices often fail to fully adjust for quality improvements embodied in capital goods. Consequently, investment is understated, capital stock growth is underestimated, and measured Total Factor Productivity (TFP) growth is biased downward.

## 3. Quantitative & Empirical Findings
Calibrating the model to postwar U.S. time series data:
- The model reproduces a substantial portion of the post-1973 productivity slowdown through the combined effects of learning costs and quality mismeasurement.
- Long-run equilibrium output and welfare expand significantly once adoption frictions subside and the learning curve matures.

## 4. Key Takeaways
The transition costs associated with adopting revolutionary general-purpose or investment-specific technologies explain why macroeconomic productivity figures may initially stagnate despite rapid technological advancement.


# Read All about It!! What Happens Following a Technology Shock?
**Author:** Michelle Alexopoulos (*American Economic Review*, 2011)

---

## 1. Introduction & Research Motivation
A foundational debate in macroeconomics concerns the short-run effect of positive technology shocks on labor input. Real Business Cycle (RBC) models predict an increase in employment, whereas standard New Keynesian models predict a contraction due to price stickiness. Empirical verification has long been contested because traditional proxies (such as Solow residuals) suffer from non-technological contamination like variable factor utilization.

## 2. Identification Strategy: Technical Publications
Alexopoulos introduces a direct, exogenous measure of technological change derived from copyrighted book metadata cataloged in the U.S. Library of Congress:
- Tracks new titles across computer science, telecommunications, and manufacturing engineering.
- Differentiates between **process innovations** (manuals, implementation guides) and **product innovations** (new software and commercial system releases).

## 3. Empirical Results
Estimating Structural Vector Autoregressions (SVARs) incorporating the publication-based innovation measures:
- Positive technology shocks induce a statistically significant, immediate expansion in gross domestic product (GDP), total factor productivity, investment, and hours worked.
- The expansionary labor response directly supports the RBC propagation mechanism, showing that technological advances do not lead to short-run employment contractions when identified via direct innovation indicators.

## 4. Conclusions
Direct measurement of innovation diffusion via technical publications resolves measurement ambiguities inherent in conventional Solow residuals, confirming positive contemporaneous comovement between labor, capital investment, and aggregate output.
