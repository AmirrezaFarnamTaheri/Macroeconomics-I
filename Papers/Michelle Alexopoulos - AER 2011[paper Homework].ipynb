## $$\small\textbf{Read All about It!! What Happens Following a Technology Shock?}$$ $$\scriptsize\textbf{by Michelle Alexopoulos - AER 2011}$$

### $\textbf{Introduction}$



The paper explores the crucial role of technical change in economic growth and fluctuations, highlighting the challenges of measuring innovative activity and technology shocks.



### $\textbf{Main Question}$


The central question is how to effectively measure technical change and its impact on economic dynamics, seeking to identify more robust indicators beyond existing proxies.


### $\textbf{Methodology and Identification}$


The author develops new book-based indicators by compiling data on new English language titles in technology fields, offering a more nuanced understanding of technological advancements. Vector Autoregression (VAR) models analyze the relationship between these measures and economic variables.


### $\textbf{Results}$


The new indicators provide a more refined perspective on technological progress, showcasing a unique relationship between technological advancements and economic productivity. Technology shocks positively impact GDP, employment, total factor productivity, and investment, with a shorter lag compared to traditional measures.


### $\textbf{Conclusion}$


The paper concludes that book-based indicators offer valuable insights into the impact of technology shocks on economic activity, emphasizing the need for precise indicators of process and product technologies to understand their impact on labor inputs and productivity. These indicators provide a more comprehensive understanding of the relationship between technological change and economic growth.
