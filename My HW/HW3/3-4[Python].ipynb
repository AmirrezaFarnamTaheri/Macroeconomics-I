```python
import numpy as np
import matplotlib.pyplot as plt

# Define the parameters
rho = 0.04
n = 0.02
g = 0.01

# Create a range for theta and alpha
theta_values = np.linspace(1, 5, 3)
alpha_values = np.linspace(1, 5, 3)

# Calculate mu_1 for each combination of theta and alpha
mu_1_values = np.zeros((len(theta_values), len(alpha_values)))
for i, theta in enumerate(theta_values):
    for j, alpha in enumerate(alpha_values):
        beta = rho - alpha*(n + g)
        mu_1_values[i, j] = 0.5 * (beta - np.sqrt(beta**2 + 4/theta*(1-alpha)/alpha*(rho + theta*g)*(rho + theta*g - alpha*(n + g))))

# Plot mu_1 as a function of theta and alpha
Theta, Alpha = np.meshgrid(theta_values, alpha_values)
plt.contourf(Theta, Alpha, mu_1_values.T, levels=25, cmap='coolwarm')
plt.colorbar()
plt.title('Adjustment Rate $\mu_1$ as a Function of $\\theta, \\alpha$')
plt.xlabel('$\\theta$')
plt.ylabel('$\\alpha$')
plt.show()
```
