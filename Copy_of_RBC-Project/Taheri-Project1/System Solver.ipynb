### RBC Project - Solving the System of Equations

```python
import sympy as sp

# Define the variables
k_t, N_t, z_t, c_t = sp.symbols('k_t N_t z_t c_t')
k_t1, z_t1, c_t1, R_t1 = sp.symbols('k_t1 z_t1 c_t1 R_t1')

# Define the equations
eq1 = sp.Eq(0.36 * k_t - 0.36 * N_t, c_t1 - 0.99 * R_t1 - z_t1 / 0.95)
eq2 = sp.Eq((0.79 / 1.06) * (c_t1 - 0.99 * R_t1) + (0.27 / 1.06) * ((k_t1 - (1 - 0.025) * k_t) / 0.025), 0.36 * k_t + 0.64 * N_t)
eq3 = sp.Eq(z_t1, 0.95 * z_t)
eq4 = sp.Eq(c_t1, 0.99 * R_t1 + c_t)


# Solve the equations for k_t, N_t, c_t, z_t
solutions = sp.solve((eq1, eq2, eq3, eq4), (k_t, N_t, c_t, z_t))

print(f"k_t: {sp.latex(solutions[k_t])}")
print(f"N_t: {sp.latex(solutions[N_t])}")
print(f"c_t: {sp.latex(solutions[c_t])}")
print(f"z_t: {sp.latex(solutions[z_t])}")

```
